# Introduction to Absolute Compensation Theory (ACT)

This notebook introduces the current Balansis runtime surface for ACT-aware arithmetic.

## What This Notebook Covers

- `AbsoluteValue` as the core signed-magnitude runtime object
- `ABSOLUTE` as the ACT additive identity in the Python runtime
- low-level compensated operations through `Operations`
- a first example of why explicit compensation matters

```python
from balansis import AbsoluteValue, Operations, ABSOLUTE, UNIT_POSITIVE, UNIT_NEGATIVE

print('Balansis ACT runtime imports loaded.')
print('ABSOLUTE:', ABSOLUTE)
print('UNIT_POSITIVE:', UNIT_POSITIVE)
print('UNIT_NEGATIVE:', UNIT_NEGATIVE)
```

## Creating `AbsoluteValue` Objects

`AbsoluteValue` stores a non-negative magnitude and a direction of `+1` or `-1`.

```python
a = AbsoluteValue(magnitude=5.0, direction=1)
b = AbsoluteValue(magnitude=3.0, direction=-1)
c = AbsoluteValue.from_float(7.5)
d = AbsoluteValue.from_float(-4.25)

print('a =', a)
print('b =', b)
print('c =', c)
print('d =', d)
print('a.is_positive() =', a.is_positive())
print('b.is_negative() =', b.is_negative())
print('ABSOLUTE.is_absolute() =', ABSOLUTE.is_absolute())
```

## Direct Arithmetic

The runtime object supports ordinary arithmetic operators, but the low-level `Operations` layer also returns explicit compensation factors.

```python
x = AbsoluteValue.from_float(8.0)
y = AbsoluteValue.from_float(-3.0)

print('x + y =', x + y)
print('x - y =', x - y)
print('-x =', -x)
print('2.5 * x =', 2.5 * x)

result, compensation = Operations.compensated_add(x, y)
print('Operations.compensated_add(x, y) ->', result, compensation)
```

## Why Compensation Matters

A classic unstable reduction can lose a meaningful residual in plain floating-point arithmetic. `Operations.sequence_sum()` keeps both the ACT result and the compensation signal visible.

```python
values = [
    AbsoluteValue.from_float(1e16),
    AbsoluteValue.from_float(1.0),
    AbsoluteValue.from_float(-1e16),
]

plain_float = sum(v.to_float() for v in values)
compensated_result, compensated_factor = Operations.sequence_sum(values)

print('Plain float sum:', plain_float)
print('Compensated ACT sum:', compensated_result)
print('Compensation factor:', compensated_factor)
```
