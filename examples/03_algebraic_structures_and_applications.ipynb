# Algebraic Structures and Runtime Applications

This notebook demonstrates the current Python runtime algebra helpers in Balansis.

## Important Scope Note

`AbsoluteGroup` and `EternityField` are runtime helper classes. They are useful for experiments and structured workflows, but they are not a replacement for the formal Lean theorem layer documented under `docs/formal/`.

```python
from balansis import AbsoluteValue, EternalRatio, AbsoluteGroup, EternityField
from balansis.algebra.absolute_group import GroupElement
from balansis.algebra.eternity_field import FieldElement

print('Runtime algebra helper imports loaded.')
```

## Additive Group over `AbsoluteValue`

The additive group helper works with `GroupElement(value=...)` wrappers.

```python
additive_group = AbsoluteGroup.additive_group()
a = GroupElement(value=AbsoluteValue.from_float(5.0))
b = GroupElement(value=AbsoluteValue.from_float(3.0))
c = GroupElement(value=AbsoluteValue.from_float(-2.0))

sum_ab = additive_group.operate(a, b)
sum_abc = additive_group.operate(sum_ab, c)
identity = additive_group.identity_element()
inverse_a = additive_group.inverse_element(a)

print('a =', a)
print('b =', b)
print('c =', c)
print('a + b =', sum_ab)
print('(a + b) + c =', sum_abc)
print('identity =', identity)
print('inverse(a) =', inverse_a)
print('a + inverse(a) =', additive_group.operate(a, inverse_a))
```

## Multiplicative Group over Non-`ABSOLUTE` Values

```python
multiplicative_group = AbsoluteGroup.multiplicative_group()
x = GroupElement(value=AbsoluteValue.from_float(4.0))
y = GroupElement(value=AbsoluteValue.from_float(0.5))

product_xy = multiplicative_group.operate(x, y)
inverse_x = multiplicative_group.inverse_element(x)

print('x =', x)
print('y =', y)
print('x * y =', product_xy)
print('inverse(x) =', inverse_x)
print('x * inverse(x) =', multiplicative_group.operate(x, inverse_x))
```

## `EternityField` Runtime Helper

The field helper wraps `EternalRatio` values inside `FieldElement(ratio=...)`.

```python
field = EternityField.rational_field()
r1 = FieldElement(ratio=EternalRatio(
    numerator=AbsoluteValue.from_float(3.0),
    denominator=AbsoluteValue.from_float(2.0),
))
r2 = FieldElement(ratio=EternalRatio(
    numerator=AbsoluteValue.from_float(5.0),
    denominator=AbsoluteValue.from_float(4.0),
))

sum_r1r2 = field.add(r1, r2)
product_r1r2 = field.multiply(r1, r2)
additive_inverse = field.additive_inverse(r1)
multiplicative_inverse = field.multiplicative_inverse(r1)

print('r1 =', r1)
print('r2 =', r2)
print('field.add(r1, r2) =', sum_r1r2)
print('field.multiply(r1, r2) =', product_r1r2)
print('field.zero() =', field.zero())
print('field.one() =', field.one())
print('field.additive_inverse(r1) =', additive_inverse)
print('field.multiplicative_inverse(r1) =', multiplicative_inverse)
```
